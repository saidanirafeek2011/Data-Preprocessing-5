**Task 2 – Data Cleaning**

```python
# 1. Rename Columns

print("Before Renaming:")
print(df.columns)

df.rename(columns={
'ID':'Visit_ID',
'No_Pation':'Patient_ID'
}, inplace=True)

print("\nAfter Renaming:")
print(df.columns)
```

```python
#2.Check Unique Values in Categorical Columns

print("\nUnique Values in Gender:")
print(df['Gender'].unique())

print("\nUnique Values in CLASS:")
print(df['CLASS'].unique())
```

```python
#3. Standardize Gender Values

df['Gender'] = df['Gender'].str.upper()

print("\nGender After Cleaning:")
print(df['Gender'].unique())
```

```python
#4. Statistical Summary

print("\nStatistical Summary")

summary = df.describe()

print(summary)
```

```python
#5. Box Plot for Numerical Columns

import matplotlib.pyplot as plt

plt.figure(figsize=(15,6))

df.select_dtypes(include=['number']).boxplot()

plt.title("Box Plot of Numerical Columns")

plt.xticks(rotation=45)

plt.show()
```

```python
#6. print("\nMissing Values")

print(df.isnull().sum())
```

```python
#7. Impute Missing Values Numerical Columns

num_cols = df.select_dtypes(include=['number']).columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Verify Missing Values Removed

print("\nMissing Values After Imputation")

print(df.isnull().sum())
```

```python
#8. Check Duplicate Rows

duplicates = df.duplicated().sum()

print("\nDuplicate Rows:", duplicates)
```

```python
# Remove Duplicates

df.drop_duplicates(inplace=True)
print("\nShape After Removing Duplicates")
print(df.shape)
```

```python
# Check most frequent value
print(df['Gender'].mode())

# Fill missing Gender values
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])

# Verify again
print(df.isnull().sum())
```

**Task 3 – Outlier Handling**

```python
#Filter Outliers in Cr (Creatinine)
#Calculate 99.5th Percentile
```

```python
lipid_cols = ['LDL','VLDL','HDL','TG','Chol']
df_before = df.copy()
df_before[lipid_cols].boxplot(figsize=(10,5))
plt.title("Before Outlier Removal")
plt.show()
```

```python
cr_threshold = df['Cr'].quantile(0.995)

print("Cr Threshold:", cr_threshold)
df = df[df['Cr'] <= cr_threshold]

print(df.shape)
```

```python
df[lipid_cols].boxplot(figsize=(10,5))
plt.title("After Outlier Removal")
plt.show()
```

```python
#Urea Column
urea_threshold = df['Urea'].quantile(0.999)
print("Urea Threshold:", urea_threshold)
df = df[df['Urea'] <= urea_threshold]
print("Shape after Urea filtering:")
print(df.shape)
```

```python
#Final Part: Lipid Columns
lipid_cols = ['LDL', 'VLDL', 'HDL', 'TG', 'Chol']

for col in lipid_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]

print("Final Shape:")
print(df.shape)
```
